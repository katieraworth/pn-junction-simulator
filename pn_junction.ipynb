```python
   %matplotlib inline
   import numpy as np
   import matplotlib.pyplot as plt
   from ipywidgets import interact, FloatLogSlider, FloatSlider
   from physics import *
   from plots import equilibrium_profiles
```

```python
def explore(NA=1e17, ND=1e16, T=300):
    x, rho, E, psi, Ec, Ev, Ei, xp, xn = equilibrium_profiles(NA, ND)
    x_um = x * 1e4
    V = np.linspace(-0.5, 0.9, 400)
    I = diode_current(V, NA, ND, T=T)

    fig, ax = plt.subplots(1, 2, figsize=(12, 4))

    ax[0].plot(x_um, Ec, label="Ec")
    ax[0].plot(x_um, Ev, label="Ev")
    ax[0].plot(x_um, Ei, "--", label="Ei")
    ax[0].axhline(0, color="k", linestyle=":", label="Ef")
    ax[0].set_xlabel("Position (µm)")
    ax[0].set_ylabel("Energy (eV)")
    ax[0].set_title("Band diagram (300 K)")
    ax[0].legend(loc="upper right", fontsize=8)

    ax[1].semilogy(V, np.abs(I))
    ax[1].set_ylim(1e-22, 10)
    ax[1].set_xlabel("Voltage (V)")
    ax[1].set_ylabel("|Current| (A)")
    ax[1].set_title(f"I-V curve ({T:.0f} K)")

    for a in ax:
        a.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()

    print(f"Vbi = {built_in_voltage(NA, ND):.3f} V  |  "
          f"W = {depletion_width(NA, ND) * 1e4:.3f} µm  |  "
          f"Emax = {max_field(NA, ND) / 1e3:.1f} kV/cm  |  "
          f"I0 = {saturation_current(NA, ND, T=T):.2e} A")


interact(explore,
         NA=FloatLogSlider(value=1e17, base=10, min=14, max=18, step=0.1, description="NA (cm⁻³)"),
         ND=FloatLogSlider(value=1e16, base=10, min=14, max=18, step=0.1, description="ND (cm⁻³)"),
         T=FloatSlider(value=300, min=250, max=400, step=5, description="T (K)"));
```
